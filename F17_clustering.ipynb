```python
import mlcroissant as mlc
import pandas as pd

dataset = mlc.Dataset('https://www.kaggle.com/datasets/bryanb/fifa-player-stats-database/croissant/download')

fifa_dataset = dataset.metadata.record_sets

fifa_df = pd.DataFrame(dataset.records(record_set=fifa_dataset[0].uuid))
pd.set_option('display.max_columns', None)
fifa_df.head()
```

```python
# Dropping all columns that we don't want to use in modeling
fifa_df=fifa_df.drop(columns=[
    'FIFA17_official_data.csv/ID',
    'FIFA17_official_data.csv/Photo',
    'FIFA17_official_data.csv/Nationality',
    'FIFA17_official_data.csv/Flag',
    'FIFA17_official_data.csv/Club',
    'FIFA17_official_data.csv/Club+Logo',
    'FIFA17_official_data.csv/Special',
    'FIFA17_official_data.csv/International+Reputation',
    'FIFA17_official_data.csv/Real+Face',
    'FIFA17_official_data.csv/Jersey+Number',
#Chose not to use position because we want best position
    'FIFA17_official_data.csv/Position',
    'FIFA17_official_data.csv/Joined',
    'FIFA17_official_data.csv/Loaned+From',
    'FIFA17_official_data.csv/Contract+Valid+Until'
    ])
```

```python
# Renaming the columns
fifa_df.rename(columns={
    'FIFA17_official_data.csv/Name': 'Player_Name', 
    'FIFA17_official_data.csv/Age': 'Player_Age',
    'FIFA17_official_data.csv/Overall': 'Player_Overall',
    'FIFA17_official_data.csv/Potential': 'Player_Potential', 
    'FIFA17_official_data.csv/Value': 'Player_Value',
    'FIFA17_official_data.csv/Wage': 'Player_Wage',
    'FIFA17_official_data.csv/Preferred+Foot': 'Preferred_Foot',
    'FIFA17_official_data.csv/Weak+Foot': 'Weak_Foot',
    'FIFA17_official_data.csv/Skill+Moves': 'Skill',
    'FIFA17_official_data.csv/Work+Rate': 'Work_Ethic',
    'FIFA17_official_data.csv/Body+Type': 'Body_Type',
    'FIFA17_official_data.csv/Position': 'Position', 
    'FIFA17_official_data.csv/Height': 'Height',
    'FIFA17_official_data.csv/Weight': 'Weight', 
    'FIFA17_official_data.csv/Crossing': 'Crossing',
    'FIFA17_official_data.csv/Finishing': 'Finishing',
    'FIFA17_official_data.csv/HeadingAccuracy': 'HeadingAccuracy',
    'FIFA17_official_data.csv/ShortPassing': 'ShortPassing',
    'FIFA17_official_data.csv/Volleys': 'Volleys',
    'FIFA17_official_data.csv/Dribbling': 'Dribbling', 
    'FIFA17_official_data.csv/Curve': 'Curve',
    'FIFA17_official_data.csv/FKAccuracy': 'FKAccuracy',
    'FIFA17_official_data.csv/LongPassing': 'LongPassing',
    'FIFA17_official_data.csv/BallControl': 'BallControl',
    'FIFA17_official_data.csv/Acceleration': 'Acceleration',
    'FIFA17_official_data.csv/SprintSpeed': 'SprintSpeed',
    'FIFA17_official_data.csv/Agility': 'Agility',
    'FIFA17_official_data.csv/Reactions': 'Reactions',
    'FIFA17_official_data.csv/Balance': 'Balance',
    'FIFA17_official_data.csv/ShotPower': 'ShotPower',
    'FIFA17_official_data.csv/GKPositioning': 'GKPositioning',
    'FIFA17_official_data.csv/GKReflexes': 'GKReflexes',
    'FIFA17_official_data.csv/Best+Position': 'Best_Position',
    'FIFA17_official_data.csv/Best+Overall+Rating': 'Best_Overall_Rating',
    'FIFA17_official_data.csv/Jumping': 'Jumping',
    'FIFA17_official_data.csv/Stamina': 'Stamina',
    'FIFA17_official_data.csv/Strength': 'Strength',
    'FIFA17_official_data.csv/LongShots': 'Long_Shots',
    'FIFA17_official_data.csv/Aggression': 'Aggression',
    'FIFA17_official_data.csv/Interceptions': 'Interceptions',
    'FIFA17_official_data.csv/Positioning': 'Positioning',
    'FIFA17_official_data.csv/Vision': 'Vision',
    'FIFA17_official_data.csv/Penalties': 'Penalties',
    'FIFA17_official_data.csv/Composure': 'Composure',
    'FIFA17_official_data.csv/Marking': 'Marking',
    'FIFA17_official_data.csv/StandingTackle': 'Standing_Tackle',
    'FIFA17_official_data.csv/SlidingTackle': 'Sliding_Tackle',
    'FIFA17_official_data.csv/GKDiving': 'GKDiving',
    'FIFA17_official_data.csv/GKHandling': 'GKHandling',
    'FIFA17_official_data.csv/GKKicking': 'GKKicking',
}, inplace=True)
```

```python
# Changing Bytes to String
byte_attributes = ['Player_Name',
                   'Player_Value',
                   'Player_Wage',
                   'Preferred_Foot',
                   'Work_Ethic',
                   'Body_Type',
                   'Height',
                   'Weight',
                   'Best_Position']

for attribute in byte_attributes:
    fifa_df[f'{attribute}'] = fifa_df[f'{attribute}'].apply(lambda x: x.decode('utf-8'))
```

```python
pd.set_option('display.max_columns', None)
fifa_df.head()
val_column = ['Player_Value', 'Player_Wage']
fifa17 = fifa_df[['Player_Name', 'Player_Age', 'Player_Potential', 'Player_Value', 'Player_Wage']]
for col in val_column:
    fifa17[col] = fifa17[col].replace({'€': '', 'K': 'e3', 'M': 'e6'}, regex=True).astype(float)

fifa17.head()
```

```python
X = fifa17[['Player_Age', 'Player_Potential', 'Player_Value', 'Player_Wage']]
X.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
inertia = []
K = range(2, 11)
for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.plot(K, inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("inertia")
plt.title("elbow Method for Optimal K")
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score
silhouette_scores = []
K = range(2, 11)
for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    score = silhouette_score(X_scaled, km.labels_)
    silhouette_scores.append(score)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(K, silhouette_scores, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Method for Optimal K")
plt.grid(True)
plt.show()
```

```python
k_chosen = 4
kmeans = KMeans(n_clusters = k_chosen, random_state=42)
kmeans.fit(X_scaled)
Labels = kmeans.labels_
```

```python
fifa17['Cluster'] = Labels

cluster_summary = fifa17.groupby('Cluster').mean(numeric_only=True)
cluster_summary
```

```python
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(
    X_scaled[:, 0], X_scaled[:, 1], X_scaled[:, 2],
    c=Labels, cmap='viridis', s=50
)
ax.set_xlabel('Age (scaled)')
ax.set_ylabel('Overall (scaled)')
ax.set_zlabel('Value (scaled)')
ax.set_title('3D Scatter Plot of FIFA 18 Player Clusters')
plt.show()
```
